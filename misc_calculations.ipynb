## Maximum distance between MIC format microphone capsules

Euclidean distance

```python

MIC_ARRAY_OFFSETS_SPH = [ # Actual MIC coordinates from Politis et Al. 2020, in spherical coordinates
    (45, 35, 0.042),
    (-45, -35, 0.042),
    (135, -35, 0.042),
    (-135, 35, 0.042),
]

p1_cart = sph2cart(MIC_ARRAY_OFFSETS_SPH[0][0], MIC_ARRAY_OFFSETS_SPH[0][1], MIC_ARRAY_OFFSETS_SPH[0][2])
p2_cart = sph2cart(MIC_ARRAY_OFFSETS_SPH[1][0], MIC_ARRAY_OFFSETS_SPH[1][1], MIC_ARRAY_OFFSETS_SPH[1][2])
p3_cart = sph2cart(MIC_ARRAY_OFFSETS_SPH[2][0], MIC_ARRAY_OFFSETS_SPH[2][1], MIC_ARRAY_OFFSETS_SPH[2][2])
p4_cart = sph2cart(MIC_ARRAY_OFFSETS_SPH[3][0], MIC_ARRAY_OFFSETS_SPH[3][1], MIC_ARRAY_OFFSETS_SPH[3][2])



print(f"Point1 to Point2: {np.sqrt(np.sum(((p1_cart-p2_cart)**2)))}")

print(f"Point1 to Point3: {np.sqrt(np.sum(((p1_cart-p3_cart)**2)))}")

print(f"Point1 to Point4: {np.sqrt(np.sum(((p1_cart-p4_cart)**2)))}")

# Thus, max dist is 0.068 meters, or 6.88 cm
```
